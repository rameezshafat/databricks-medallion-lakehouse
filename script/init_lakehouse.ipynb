# Lakehouse Setup
One-time setup of the Unity Catalog objects the pipeline writes to: one schema per medallion layer (`bronze`, `silver`, `gold`) and a volume that holds the raw CSV files.

Run this once before any bronze notebook.

# Select Catalog

```python
%sql
-- Every schema, table and volume in this project lives in the `workspace` catalog.
USE CATALOG workspace;
```

# Create Lakehouse Schemas

```python
%sql
-- One schema per medallion layer. IF NOT EXISTS makes this notebook safe to re-run.
CREATE SCHEMA IF NOT EXISTS bronze
COMMENT 'Bronze layer: raw ingested data';

CREATE SCHEMA IF NOT EXISTS silver
COMMENT 'Silver layer: cleaned and transformed data';

CREATE SCHEMA IF NOT EXISTS gold
COMMENT 'Gold layer: business-ready data';
```

#Create Volume


```python
%sql
-- Raw CSVs are uploaded here under source_crm/ and source_erp/.
-- The bronze notebooks read them from /Volumes/workspace/bronze/raw_sources/...
CREATE VOLUME IF NOT EXISTS workspace.bronze.raw_sources
COMMENT 'Volume for raw source files (CSV)';
```
