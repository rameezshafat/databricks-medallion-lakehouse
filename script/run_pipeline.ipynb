## Pipeline Entry Point
Runs the whole lakehouse end to end: bronze ingestion, then the silver and gold orchestration notebooks (gold ends with the data quality checks).

This is the single notebook to schedule as a Databricks Job. Run `init_lakehouse` once beforehand.

```python
# Each layer reads only what the previous one wrote, so they run strictly in order.
# If any child notebook fails (including a failed quality check), dbutils.notebook.run raises and the run stops.
layers = [
    "./bronze/bronze_layer_(improved)",
    "./silver/silver_orchestration",
    "./gold/gold_orchestration"
]

for nb in layers:
    print(f"Running {nb}")
    dbutils.notebook.run(nb, timeout_seconds=0)
```
