# Bronze Ingestion (config-driven)
Loads all six raw CSVs (3 CRM, 3 ERP) into Delta tables in `workspace.bronze`.

Sources are declared once in `INGESTION_CONFIG`, replacing the one-cell-per-file approach in `bronze_layer(basic)`.

#Define Ingestion Configuration

```python
# One entry per raw file: source system, path inside the raw_sources volume, and target table in workspace.bronze.
# Adding a new source file means adding an entry here; the ingestion loop below doesn't change.
INGESTION_CONFIG = [
    {
        "source": "crm",
        "path": "/Volumes/workspace/bronze/raw_sources/source_crm/cust_info.csv",
        "table": "crm_cust_info"
    },
    {
        "source": "crm",
        "path": "/Volumes/workspace/bronze/raw_sources/source_crm/prd_info.csv",
        "table": "crm_prd_info"
    },
    {
        "source": "crm",
        "path": "/Volumes/workspace/bronze/raw_sources/source_crm/sales_details.csv",
        "table": "crm_sales_details"
    },
    {
        "source": "erp",
        "path": "/Volumes/workspace/bronze/raw_sources/source_erp/CUST_AZ12.csv",
        "table": "erp_cust_az12"
    },
    {
        "source": "erp",
        "path": "/Volumes/workspace/bronze/raw_sources/source_erp/LOC_A101.csv",
        "table": "erp_loc_a101"
    },
    {
        "source": "erp",
        "path": "/Volumes/workspace/bronze/raw_sources/source_erp/PX_CAT_G1V2.csv",
        "table": "erp_px_cat_g1v2"
    }
]
```

#Ingest Files into Bronze Tables

```python
%python

for item in INGESTION_CONFIG:
    print(f"Ingesting {item['source']} → workspace.bronze.{item['table']}")

    # Bronze keeps the data as delivered: header row + inferred schema, no cleaning.
    df = (
        spark.read
             .option("header", "true")
             .option("inferSchema", "true")
             .csv(item["path"])
    )

    # Full overwrite on every run (no incremental load); Delta format is set explicitly.
    (
        df.write
          .mode("overwrite")
          .format("delta")
          .saveAsTable(f"workspace.bronze.{item['table']}")
    )
```
