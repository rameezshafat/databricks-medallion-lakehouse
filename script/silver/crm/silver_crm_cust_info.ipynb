# Silver: CRM Customers
`bronze.crm_cust_info` → `silver.crm_customers`

Cleans the CRM customer master: trims text, expands marital-status and gender codes, drops rows without a customer ID, and renames columns to business names.

#Initialization

```python
import pyspark.sql.functions as F
from pyspark.sql.types import StringType
from pyspark.sql.functions import trim, col
```

# Read Bronze table

```python
df = spark.table("workspace.bronze.crm_cust_info")
```

#Silver Transformations

##Trimming

```python
# Trim every string column found in the schema, so this cell keeps working if columns are added.
# The raw CRM names have stray spaces, e.g. ' Jon' and 'Yang '.
for field in df.schema.fields:
    if isinstance(field.dataType, StringType):
        df = df.withColumn(field.name, trim(col(field.name)))
```

##Normalization

```python
# Expand single-letter codes into readable labels. Anything unrecognised (including NULL) becomes 'n/a',
# so gold never has to handle NULLs in these columns.
df = (
    df
    .withColumn(
        "cst_marital_status",
        F.when(F.upper(F.col("cst_marital_status")) == "S", "Single")
         .when(F.upper(F.col("cst_marital_status")) == "M", "Married")
         .otherwise("n/a")
    )
    .withColumn(
        "cst_gndr",
        F.when(F.upper(F.col("cst_gndr")) == "F", "Female")
         .when(F.upper(F.col("cst_gndr")) == "M", "Male")
         .otherwise("n/a")
    )
)
```

##Remove Records with Missing Customer ID

```python
# A customer without cst_id can't be joined to sales or ERP data, so the row is dropped.
# TODO: cst_id isn't de-duplicated yet; the source has a few repeated IDs (see README known gaps).
df = df.filter(col("cst_id").isNotNull())
```

## Renaming Columns

```python
# CRM source names (cst_*) -> business names used in silver and gold.
# customer_number (cst_key) is the key the ERP tables are aligned to.
RENAME_MAP = {
    "cst_id": "customer_id",
    "cst_key": "customer_number",
    "cst_firstname": "first_name",
    "cst_lastname": "last_name",
    "cst_marital_status": "marital_status",
    "cst_gndr": "gender",
    "cst_create_date": "created_date"
}
for old_name, new_name in RENAME_MAP.items():
    df = df.withColumnRenamed(old_name, new_name)
```

#Writing Silver Table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.silver.crm_customers")
```
