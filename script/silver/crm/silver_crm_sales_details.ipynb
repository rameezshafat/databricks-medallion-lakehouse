# Silver: CRM Sales
`bronze.crm_sales_details` → `silver.crm_sales`

Parses integer-encoded dates, repairs invalid prices and sales amounts, and renames columns.

#Initialization

```python
import pyspark.sql.functions as F
from pyspark.sql.types import StringType, DateType
from pyspark.sql.functions import col, trim, length
```

# Read Bronze table

```python
df = spark.table("workspace.bronze.crm_sales_details")
```

#Silver Transformations

##Trimming

```python
# Trim every string column found in the schema, so this cell keeps working if columns are added.
for field in df.schema.fields:
    if isinstance(field.dataType, StringType):
        df = df.withColumn(field.name, trim(col(field.name)))
```

##Cleaning Dates

```python
# Dates arrive as yyyyMMdd integers (e.g. 20101229). A 0, or any value that isn't 8 digits,
# is treated as missing (NULL); valid values are parsed into real DATE columns.
df = (
    df
    .withColumn(
        "sls_order_dt",
        F.when(
            (col("sls_order_dt") == 0) | (length(col("sls_order_dt")) != 8),
            None
        ).otherwise(F.to_date(col("sls_order_dt").cast("string"), "yyyyMMdd"))
    )
    .withColumn(
        "sls_ship_dt",
        F.when(
            (col("sls_ship_dt") == 0) | (length(col("sls_ship_dt")) != 8),
            None
        ).otherwise(F.to_date(col("sls_ship_dt").cast("string"), "yyyyMMdd"))
    )
    .withColumn(
        "sls_due_dt",
        F.when(
            (col("sls_due_dt") == 0) | (length(col("sls_due_dt")) != 8),
            None
        ).otherwise(F.to_date(col("sls_due_dt").cast("string"), "yyyyMMdd"))
    )
)
```

##Sales and Price Corrections

```python
# About 35 rows have inconsistent money columns: missing or negative prices, missing/zero/negative sales,
# and sales that don't equal quantity * price. Repair in two steps.

# 1) Price. A negative price is a sign error (e.g. -769 with sales 769), so take its absolute value.
#    A missing or zero price is derived from sales / quantity (guarding against division by zero).
df = df.withColumn(
    "sls_price",
    F.when(col("sls_price") < 0, F.abs(col("sls_price")))
     .when(
        col("sls_price").isNull() | (col("sls_price") == 0),
        F.when(col("sls_quantity") != 0, col("sls_sales") / col("sls_quantity"))
    )
     .otherwise(col("sls_price"))
)

# 2) Sales. Missing, non-positive, or not equal to quantity * price -> recompute it.
#    When the two disagree, price is trusted: it's the unit-level value, and sales is derived from it.
expected_sales = col("sls_quantity") * col("sls_price")
df = df.withColumn(
    "sls_sales",
    F.when(
        col("sls_sales").isNull() | (col("sls_sales") <= 0) | (col("sls_sales") != expected_sales),
        expected_sales
    ).otherwise(col("sls_sales"))
)
```

## Renaming Columns

```python
# product_number and customer_id are the keys gold uses to look up dim_products and dim_customers.
RENAME_MAP = {
    "sls_ord_num": "order_number",
    "sls_prd_key": "product_number",
    "sls_cust_id": "customer_id",
    "sls_order_dt": "order_date",
    "sls_ship_dt": "ship_date",
    "sls_due_dt": "due_date",
    "sls_sales": "sales_amount",
    "sls_quantity": "quantity",
    "sls_price": "price"
}
for old_name, new_name in RENAME_MAP.items():
    df = df.withColumnRenamed(old_name, new_name)
```

#Writing Silver Table

```python
df.write.mode("overwrite").format("delta").saveAsTable("workspace.silver.crm_sales")
```
